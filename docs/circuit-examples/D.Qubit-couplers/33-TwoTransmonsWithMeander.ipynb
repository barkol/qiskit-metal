### Create the design in Metal
Setup a design of a given dimension. Dimensions will be respected in the design rendering. <br>
The chip is centred on the origin: it spans ±size_x/2 and ±size_y/2.

```python
design = designs.DesignPlanar({}, True)
design.chips.main.size["size_x"] = "4mm"
design.chips.main.size["size_y"] = "6mm"

gui = MetalGUI(design)
```

Create two transmons coupled by a meandered bus, placed symmetrically about the chip centre.

`total_length` sets the bus frequency. The 5 mm line drawn here is an open-open λ/2 resonator near 12 GHz for the default 10/6 um CPW on silicon, well above the transmons, so this is a layout example; a bus near 6.5 GHz needs about 9.4 mm (somewhat less once the coupling pads load it).

```python
from qiskit_metal.qlibrary.qubits.transmon_pocket import TransmonPocket
from qiskit_metal.qlibrary.tlines.meandered import RouteMeander
```

```python
TransmonPocket.get_template_options(design)
```

```python
RouteMeander.get_template_options(design)
```

```python
q1 = TransmonPocket(
    design,
    "Q1",
    options=dict(
        pos_x="-1.0 mm",
        pad_width="425 um",
        pocket_height="650um",
        connection_pads=dict(readout=dict(loc_W=+1, loc_H=+1, pad_width="200um")),
    ),
)

q2 = TransmonPocket(
    design,
    "Q2",
    options=dict(
        pos_x="+1.0 mm",
        pad_width="425 um",
        pocket_height="650um",
        connection_pads=dict(readout=dict(loc_W=-1, loc_H=+1, pad_width="200um")),
    ),
)
```

```python
coupler_options = Dict(
    pin_inputs=Dict(
        start_pin=Dict(component="Q1", pin="readout"),
        end_pin=Dict(component="Q2", pin="readout"),
    ),
    fillet="99.9um",
    total_length="5mm",
    lead=Dict(start_straight="200um"),
)


bus = RouteMeander(design, "coupler", options=coupler_options)

gui.rebuild()
gui.autoscale()
```

```python
# Get a list of all the qcomponents in QDesign and then zoom on them.
all_component_names = design.components.keys()

gui.zoom_on_components(all_component_names)
```
